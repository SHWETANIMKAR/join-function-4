```python
import pandas as pd 
# Table 1: Movie details
movies = pd.DataFrame({
    "Genre": ["Drama", "Action", "Comedy", "Thriller"],
    "Year": [2016, 2023, 2009, 2022]
}, index=["Dangal", "Jawan", "3 Idiots", "Drishyam"])

# Table 2: Movie ratings
ratings = pd.DataFrame({
    "Rating": [8.4, 7.0, 8.4, 8.2],
    "Votes": [2000, 1500, 1800, 1200]
}, index=["Dangal", "Jawan", "PK", "Drishyam"])

print("MOVIES TABLE")
print(movies)

print("\nRATINGS TABLE")
print(ratings)
```

```python
import pandas as pd
movies = pd.DataFrame({
    "Genre": ["Drama", "Action", "Comedy", "Thriller"],
    "Year": [2016, 2023, 2009, 2022]
}, index=["Dangal", "Jawan", "3 Idiots", "Drishyam"])

# Report B: Rating information available
ratings = pd.DataFrame({
    "Rating": [8.4, 7.0, 8.2, 7.8],
    "Votes": [2000, 1500, 1200, 900]
}, index=["Dangal", "Jawan", "Drishyam", "Pathaan"])

print("MOVIE DETAILS")
print(movies)

print("\nRATING DETAILS")
print(ratings)
```
